## Perceptron


```python
import numpy as np
from sklearn.datasets import load_iris
from sklearn.linear_model import Perceptron
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
# Load iris data
iris = load_iris()

# We want to set up a binary classifier for setosa
# using petal length and petal width

# First check what iris contains
print(iris.keys())

# Check which columns are the relevant features
print(iris.feature_names) # columns 2 and 3

# Next, see which target represents setosa
print(iris.target_names) # column 0
```

```python
# Select petal length and petal width as features
X = iris.data[:, (2, 3)]

# Set target as setosa
y = (iris.target == 0).astype(int)

# Train/Test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=52, shuffle=True, stratify=y)

# Set up perceptron classifier
per_clf = Perceptron()
per_clf.fit(X_train, y_train)

y_pred = per_clf.predict(X_test)
score = accuracy_score(y_test, y_pred)

print(f"Accuracy: {score}")
```
