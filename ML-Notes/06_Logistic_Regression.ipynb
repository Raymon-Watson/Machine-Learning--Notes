# 06 Logistic Regression

Similarly to Linear Regression, *Logistic Regression* computes a linear combination of input features of the form $x^T \theta$, but then passes it through a logistic function:
$$
\sigma(t) = \frac{1}{1 + \exp(-t)}
$$

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10,10,100)
y = 1/(1 + np.exp(-x))
plt.title("Logistic Function")
plt.xlabel("t")
plt.ylabel("\u03C3(t)")
plt.plot(x,y)
plt.show()
```

Through this, we define the Logistic Regression model **estimated probability**:
$$
\hat{p} = h_{\theta}(x) = \sigma(x^T \theta)
$$
