# Linear Regression Notes

In this notebook, I will give a general overview of Linear Regression with application in Python. This will be in the context of data analysis, so the presentation will additionally contain details on methods for extracting information from data using pandas.

This notebook will therefore additionally contain relevant code that can be repurposed for future linear regression problems.

A general overview of the pipeline used for applying linear regression to data is as follows:
- Import data to pandas DataFrame
- Exploratory data analysis
    - Examine target variable
    - Plot relevant data to find trends
    - Examine correlations
- Choose relevant X and y
- Train/test split
- Apply LinearRegression() from scikit-learn
- .fit() to training data
- .predict() to test data
- MAE/RMSE/$R^2$
- Plot predictions/residuals
- Interpret results

## List of things to understand:
- Model:
$$ \hat y = \beta_0+\beta_1x_1+\cdots+\beta_px_p $$
- Residuals and errors
- Mean squared error
$$ \text{MSE}=\frac1N\sum_i(y_i-\hat y_i)^2 $$
- Ordinary least squares
- Geometric interpretation of least squares
- Normal equations conceptually
- Gradient descent conceptually
- Interpretation of coefficients
- Continuous vs categorical predictors
- Dummy variables / one-hot encoding
- Interaction terms
- Polynomial regression
- Linearity assumption
- Independence of errors
- Homoscedasticity
- Normally distributed errors and when that matters
- Multicollinearity
- Outliers and influential observations
- \(R^2\) and adjusted \(R^2\)
- Training vs test error
- Bias–variance tradeoff
- Be able to explain why minimizing squared error produces the fitted regression coefficients.
