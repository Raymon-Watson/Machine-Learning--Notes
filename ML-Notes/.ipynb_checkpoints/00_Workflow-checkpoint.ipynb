## 1. Frame the problem and look at the big picture

**Summary:** look carefully at the problem you want to solve, including current methods for solving the problem. Gather information from all relevant sources.


This is an extremely important step to start with, as we are not solely focused on building a model, instead we want to carefully select our modelling methods to serve the big picture, or find a solution to a problem that has been presented to us.

**Note on Pipelines:** pipelines is terminology for a sequence of data processing components. Typically in ML projects there is a lot of data to manipulate and many data transformations to apply, hence the overall project is constituted of various pipelines, often managed by distinct teams.

If a current solution exists, it is important to find out what it looks like, as this will give insights into potential methods which can be used to solve the problem. Overall, this initial stage is information gathering from relevant sources.

### Framing the problem
**Summary:** Explicitly frame the problem you are tackling, this will help define the method utilized in your solution.

*Framing the problem* consits of defining exactly what you are looking to do when designing your system. This is an essential first step, as correctly framing the problem means that you will not waste time applying methods that are not relevant, or answering questions that are not useful. Here, we answer questions such as:
- Is this supervised or unsupervised? Or is it reinforcement learning?
- Is this a classification task? Or a regression task? If so, what type of classification/regression task?
- Should we use batch learning or another method?
Answers to these questions should be used throughout the project to keep you on track and to define exactly what sort of ML methods are used.

### Selecting a performance measure
**Summary:** Select a relevant performance measure, commonly RMSE, but may also depend on the size of the data set and the weight you give to errors.

Next, we select a performance measure. For linear regression, this is commonly given by the *Root Mean Square Error (RMSE)*:
$$
\mathrm{RMSE}(\mathbf{X},h) = \sqrt{ \frac{1}{m} \sum_{i=1}^m \left(h(\mathbf{x}^{(i)}) - y^{(i)} \right)^2}
$$
**Note on notation:**
- $m$ is the number of instances in the dataset
- $\mathbf{x}^{(i)}$ is a vector of all the feature values of the $i^{\mathrm{th}}$ instance of the dataset, and $y^{(i)}$ is its label
- $\mathbf{X}$ is a matrix containing all feature values (excluding labels) of all instances in the dataset, where there is one row per instance, and the $i^{\mathrm{th}}$ row is equal to the transpose of $\mathbf{x}^{(i)}$
- $h$ is your system's prediction function, also called a *hypothesis*. When your system is given an instance's feature vector $\mathbf{x}^{(i)}$, it outputs a predicted value $\hat{y}^{(i)} = h(\mathbf{x}^{(i)})$
- $\mathrm{RMSE}(\mathbf{X},h)$ is the cost function measured on the set of examples using your hypothesis $h$.

Note that the RMSE is not the only performance measure for regression tasks (though it is often preferred). Another example of a performance measure is the *Mean Absolute Error (MAE)*, also called the average deviation
$$
\mathrm{MAE}(\mathbf{X},h) = \frac{1}{m} \sum_{i=1}^m \big| h(\mathbf{x}^{(i)}) - y^{(i)}\big|
$$
As a note, the above consist of two distinct ways of measuring distnaces between two vectors, and are part of the $l_k$ family of norms:
$$
||\mathbf{v}||_k = (|v_0|^k + \dots + |v_n|^k)^{1/k}
$$
The higher the norm index, the more it focuses on large values and neglects small ones, meaning e.g. that the RMSE is more sensitive to outliers than the MAE, however most data is assumed to possess Gaussian noise, for which the RMSE performs extremely well and is generally preferred.

### Create a test set
**Summary:** Use scikit-learn to split your data into training and testing sets at this point. Care needs to be taken to make sure these remain distinct, and are representative of the various strata of your population.


At this point, we want to split the data into training and testing with an 80/20 split (generally). To do this we will use train_test_split from sklearn.model_selection:
```python
train_set, test_set = train_test_split(data, test_size=0.2, random_state=42)
```
- data: the data you are splitting (i.e. your DataFrame)

- random_state: the seed for repeatable results (assuming the data set is at a fixed size)

#### Take care with train/test splits
If we add new data then the train/test split will change, even with a fixed seed. This means we may test data previously in the training set. The simplest solution is to **save the split**, which is viable for data sets that aren't too large.

Another way would be to use an identifier to keep the splitting stable. We won't go into much detail on this now, but come back to this later when it is useful.


Additionally, *random sampling isn't always representative.* An example of this would be sampling from a housing dataset where the neighborhood quality has a distinct distribution. If we simply randomly select, we might inadvertently bias, especially for smaller datasets. **Stratified sampling** is a solution to this, and it can be easily done using scikit-learn. If we have an important variable, "variable_1", that has a distinct stratification, we can use:
```python
train_set, test_set = train_test_split(
    data, 
    test_size=0.2, 
    random_state=42, 
    stratify=df["variable_1"]
)
```
Note that there are a variety of other ways of doing this, and we can come back to this later.

### Feature scaling

ML algorithms tend to work poorly when numerical data have very different scales. Two common ways to fix this are:
- *min-max scaling*: often called normalization, is done by subtracting the min value and dividing by max minus the min. Scikit-Learn has MinMaxScaler to do this.
- *standardization*: subtracts the mean value and then divides by the standard deviation (typical standardization for gaussian distributions). Scikit-Learn has StandardScaler to do this.


### Transformation pipelines

It is essential that our transformations are done in the right order. Scikit-Learn has the Pipeline class to help with such sequences of transformations. This takes in data for preprocessing and can finish with a final predictor.

## 5. Explore many different models and shortlist the best ones

Having completed the previous steps, training a model becomes much simpler than it would otherwise be. We have observed which features are correlated with our target variable, we have identified possible new features, and scaled our variables. We are ready to train our models.

### Training and evaluating on the training set

We focus here on linear regression, though much of this is simply extended to other methods. We call our prepared training data 'df_prepared', which contains all the predictors, and 'df_labels', which contains the labels (i.e. the target variable) that we have cleaned and prepared.

To train our linear regression model, all we need to do is:
```python
from sklearn.linear_model import LinearRegression

lin_reg = LinearRegression()
lin_reg.fit(df_prepared, df_labels)
```
and we have trained our model!

We can then measure our regression model's RMSE, which is the metric we decided to use at the start,
```python
from sklearn.metrics import mean_squared_error
df_predictions = lin_reg.predict(df_prepared)
lin_mse = mean_squared_error(df_labels, df_predictions)
lin_rmse = np.sqrt(lin_mse)
print(lin_rmse)
```
(Note: this is a metric performed on the training set, so it is an in-sample metric.)

As an example of a more complex model, let's try training a decision tree regressor:
```python
# Training
from sklearn.tree import DecisionTreeRegressor
tree_reg = DecisionTreeRegressor()
tree_reg.fit(df_prepared, df_labels)

# Metrics
df_predictions = tree_reg.predict(df_prepared)
tree_mse = mean_squared_error(df_labels, df_predictions)
tree_rmse = np.sqrt(tree_mse)
print(tree_rmse)
```
We can see from this example that fitting a more complex model takes the exact same form, making our job of training our model much easier!


Another method we could use is Random Forest regression. Again, fitting this takes a familiar form:
```python
# Training
from sklearn.ensemble import RandomForestRegressor
forest_reg = RandomForestRegressor()
forest_reg.fit(df_prepared, df_labels)

# Metrics
df_predictions = forest_reg.predict(df_prepared)
forest_mse = mean_squared_error(df_labels, df_predictions)
forest_rmse = np.sqrt(forest_mse)
print(forest_rmse)
```

### Cross-validation

At this point we should be wary of under- and over-fitting our data, we can check by inspection, or later by testing on the test set. But for now, we will try a better evaluation using cross-validation.

To do this, we can again use train_test_split() to split the *training set* into a smaller training set and a validation set. To do this in the easiest way possible, we can use Scikit-Learn's *K-fold cross-validation* feature, which splits the training set into 10 distinct subsets (called *folds*) and then trains and evaluates the model 10 times, picking a different fold for each evaluation every time and training on the other 9 folds. This gives an array of 10 evaluation scores.

We give our first example for Linear Regression:
```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(lin_reg, df_prepared, df_labels,
                         scoring="neg_mean_squared_error", cv=10)
lin_rmse_scores = np.sqrt(-scores)
```
(Note: this function actually expects a utility function, which means the higher the score the better, which is why we have the negative values in the above).

To look at the results, we can write up a simple function:
```python
def display_scores(scores):
    print("Scores:", scores)
    print("Mean:", scores.mean())
    print("Standard deviation:", scores.std())

display_scores(lin_rmse_scores)
```

The extension to other trained models is simple. In fact, we can use the exact same method to train a variety of models, which we can compare amongst, making sure to check the training error and the validation error, mindful of overfitting. From here we can **shortlist two to five promising models**.


## 6.  Fine-tune your models
Having a shortlist of promising models, we now turn to fine-tuning them. In this section we list a few ways of accomplishing this.

### Grid Search
If we are looking at tuning a small number of hyperparameters, using a grid search is often optimal.

Often we will want to tweak our hyperparameters until the best combination of values is found. To do this, we can use Scikit-Learn's *GridSearchCV*. This method takes our specified hyperparameters, along with values we want to try out, and uses cross-validation to evaluate all the possible combinations of hyperparameter values.

Let's look at an example for the Random Forest Regressor introduced above:
```python
param_grid = [
    {'n_estimators': [3, 10, 30], 'max_features': [2, 4, 6, 8]},
    {'bootstrap': [False], 'n_estimators': [3, 10], 'max_features': [2,3,4]}
]

forest_reg = RandomForestRegressor()

grid_search = GridSearchCV(forest_reg, param_grid, cv=5,
                           scoring='neg_mean_squared_error',
                           return_train_score=True)

grid_search.fit(df_prepared, df_labels)
```
We now go over what exactly the above has done in detail.

#### param_grid
This defines the set of hyperparameter combinations that GridSearchCV will try. Because we have two dictionaries, what it will do is try all combinations from the first dictionary, then try all combinations from the second dictionary.

The *first dictionary* consists of 
- n_estimators: controls the number of trees in the random forest (i.e. 3 trees, 10 trees, and 30 trees)
- max_features: controls how many input features each decision tree is allowed to consider when looking for the best split (i.e. 2 features, 4 features, 6 features, and 8 features)

These combine to $3 \times 4 = 12$ different configurations.

Additionally, since we haven't specified bootstrap in the first dictionary, it defaults to True for the first search.


#### forest_reg
This obviously just chooses the type of ML model we are fitting to the data. In combination with what is specified in the param_grid above, the GridSearchCV will effectively try out searches like
```python
RandomForestRegressor(
    bootstrap=False,
    n_estimator=3,
    max_features=2
)
```
and so on...

#### GridSearchCV

This is us setting up the grid search properly, which will be executed in the next step. 
- forest_reg says we are performing forest regression
- param_grid inputs the parameter grid defined above
- cv=5 means using 5-fold cross-validation
- scoring="neg_mean_squared_error" dictates the method by which the quality of the model is assessed
- return_train_score=True tells the search to record the scores on the training portion of each fold, and not just the validation result

*Note:* the results are stored inside:
```python
grid_search.cv_results_
```

(Note: If you have no idea what value a hyperparameter should have, a simple approach is to try out consecutive powers of 10, or smaller for a finer grain.)

#### grid_search.fit()

This runs the grid search using all the input data defined above.


#### Output

We can look at the *best combination of parameters*:
```python
grid_search.best_params_
```
(Note: if our best features return as the maximum values specified, it would be best to search again with higher values, as the score may improve further.)

Or we can get the best estimators directly:
```python
grid_search.best_estimator_
```
And also look at the evaluation scores:
```python
cvres = grid_search.cv_results_
for mean_score, params in zip(cvres["mean_test_scores"], cvres["params"]):
    print(np.sqrt(-mean_score), params)
```


### Randomized Search


When our hyperparameter space is large, we often prefer to use RandomizedSearchCV. This is used in much the same way as GridSearchCV, but instead of trying out all possible combinations, it evaluates a given number of random combinations. There are two main benefits:
- It explores the space quite well, since 1,000 iterations will explore 1,000 different values for each hyperparameter
- By setting the number of iterations we have more control over the computing budget we want to allocate to our hyperparameter search
