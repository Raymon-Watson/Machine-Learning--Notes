# Basic Regression


```python
import numpy as np
import matplotlib.pyplot as plt
```

## Linear Regression

Written concisely, the linear regression prediction formula is given as
$$
\hat{y} = h_{\theta} (x) = \theta \cdot x
$$
where:
- $\theta$ is the parameter vector (containing the bias term, $\theta_0$)
- $x$ is the feature vector (containing $x_0=1$)
- $h_\theta$ is the hypothesis function, using the model parameters $\theta$.

Note: we will write the vectors $\theta$ and $x$ without using bold, denoting their elements when required by, e.g., $\theta_i$.


To train this model, it is common (and in most models this is also true) to use the RMSE. In practice we use the MSE, since it leads to the same result:
$$
\mathrm{MSE}(X,h_{\theta}) = \frac{1}{m} \sum_{i=1}^m \left( \theta^T x^{(i)} - y^{(i)} \right)^2
$$
where $X$ is our training set, containing $m$ features represented as vectors $x^{(i)}$ and their associated values $y^{(i)}$.

### The Normal Equation

The MSE has a closed form solution, known as the *normal equation*:
$$
\hat{\theta} = (X^T X)^{-1} X^T y
$$
Here we give a brief example of how this works on data:

```python
Xmax = 3       # Maximum X value
intercept = 4  # intercept of slope
slope = 3      # gradient of slope
randScale = 2  # scale of random data

# Generate Data
X = Xmax * np.random.rand(100, 1)
y = intercept + slope * X + randScale * np.random.randn(100,1)

# Normal Equation
X_b = np.c_[np.ones((100, 1)), X] # adds x0 = 1 to each data point
theta_best = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
print(theta_best)

# Plot data and linear fit
plt.scatter(X,y)
plt.plot
plt.show()
```
