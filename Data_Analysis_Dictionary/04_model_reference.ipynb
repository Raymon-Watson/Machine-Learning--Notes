### Quick Rules of Thumb

- **Need a simple baseline:** Linear Regression or Logistic Regression
- **Need interpretability:** Linear/Logistic Regression, Ridge, Lasso, shallow Decision Trees
- **Nonlinear tabular data:** Random Forest or Gradient Boosting
- **Strongest general tabular performance:** XGBoost, LightGBM, or CatBoost
- **Many categorical features:** CatBoost is particularly worth trying
- **Small nonlinear dataset:** SVM/SVR
- **High-dimensional sparse data:** Logistic Regression or Linear SVM
- **Very large tabular dataset:** LightGBM or XGBoost
- **Need robustness to outliers:** Huber or RANSAC
- **Need feature selection:** Lasso or Elastic Net
- **Need uncertainty estimates:** Gaussian Process Regression or Bayesian Ridge
- **Distance-based models:** KNN and SVM generally require scaling
- **Tree-based models:** scaling is generally unnecessary
- **Neural networks:** scaling is usually essential

### 1.1 Linear Regression

#### Overview
Fits a linear relationship between the input features and a continuous target by minimizing the sum of squared residuals.

Good as a **baseline regression model** and particularly useful when interpretability is important and relationships are approximately linear.

**Scaling is not required for prediction, but is useful when comparing coefficient magnitudes.**

#### Syntax

```python
from sklearn.linear_model import LinearRegression

lin_reg = LinearRegression(
    fit_intercept=True
)
```

#### Parameters

| Parameter | Effect |
|---|---|
| `fit_intercept` | Whether an intercept is estimated |
| `positive` | Forces coefficients to be non-negative |

#### Tuning strategy

There is generally very little to tune.

Instead, concentrate on:
- feature engineering
- transformations
- checking residuals
- interaction terms
- polynomial features
- regularized alternatives such as Ridge/Lasso

#### Advantages
- Simple and fast
- Highly interpretable
- Strong baseline
- Works well when relationships are genuinely linear

#### Disadvantages
- Cannot naturally capture nonlinear relationships
- Sensitive to outliers
- Multicollinearity can make coefficients unstable
- Assumes additive relationships

#### Additional Notes

Ordinary least squares minimizes:

$$
\mathrm{RSS}=\sum_{i=1}^{n}(y_i-\hat y_i)^2
$$


### 1.2 Ridge Regression

#### Overview
Linear regression with **L2 regularization**, which penalizes large coefficients.

Particularly useful when features are correlated or ordinary linear regression is overfitting.

**Scaling strongly recommended.**

#### Syntax

```python
from sklearn.linear_model import Ridge

ridge = Ridge(
    alpha=1.0
)
```

#### Parameters

| Parameter | Effect |
|---|---|
| `alpha` | Strength of L2 regularization |
| `solver` | Optimization algorithm |

#### Tuning strategy

```python
param_grid = {
    "alpha": [0.001, 0.01, 0.1, 1, 10, 100]
}
```

If overfitting:
- increase `alpha`

If underfitting:
- decrease `alpha`

#### Advantages
- Reduces overfitting
- Handles correlated features well
- Stable compared with ordinary least squares
- Retains all features

#### Disadvantages
- Less interpretable after regularization
- Does not perform feature selection
- Still assumes approximately linear relationships

#### Additional Notes

Ridge minimizes:

$$
\mathrm{RSS}+\alpha\sum_j\beta_j^2
$$

### 1.3 Lasso Regression

#### Overview
Linear regression with **L1 regularization**.

Lasso can force some coefficients exactly to zero, meaning it can perform automatic feature selection.

**Scaling strongly recommended.**

#### Syntax

```python
from sklearn.linear_model import Lasso

lasso = Lasso(
    alpha=1.0,
    max_iter=10000
)
```

#### Parameters

| Parameter | Effect |
|---|---|
| `alpha` | Strength of L1 regularization |
| `max_iter` | Maximum optimization iterations |
| `tol` | Convergence tolerance |

#### Tuning strategy

```python
param_grid = {
    "alpha": [0.0001, 0.001, 0.01, 0.1, 1, 10]
}
```

Higher `alpha`:
- stronger regularization
- more coefficients driven to zero

#### Advantages
- Performs feature selection
- Can reduce overfitting
- Useful for high-dimensional data
- Produces sparse models

#### Disadvantages
- Can behave unpredictably with strongly correlated features
- Can discard useful variables
- Linear relationships only

#### Additional Notes

Lasso minimizes:

$$
\mathrm{RSS}+\alpha\sum_j|\beta_j|
$$


### 1.4 Elastic Net

#### Overview
Combines L1 regularization from Lasso with L2 regularization from Ridge.

Useful when there are **many correlated features** and some degree of feature selection is desired.

**Scaling strongly recommended.**

#### Syntax

```python
from sklearn.linear_model import ElasticNet

elastic = ElasticNet(
    alpha=1.0,
    l1_ratio=0.5,
    max_iter=10000
)
```

#### Parameters

| Parameter | Effect |
|---|---|
| `alpha` | Overall regularization strength |
| `l1_ratio` | Balance between L1 and L2 penalties |
| `max_iter` | Maximum optimization iterations |

#### Tuning strategy

```python
param_grid = {
    "alpha": [0.001, 0.01, 0.1, 1, 10],
    "l1_ratio": [0.1, 0.25, 0.5, 0.75, 0.9]
}
```

`l1_ratio → 1` approaches Lasso.

`l1_ratio → 0` approaches Ridge.

#### Advantages
- Handles correlated features
- Can perform feature selection
- More flexible than Ridge or Lasso individually

#### Disadvantages
- Additional hyperparameter
- Requires scaling
- Still limited to linear relationships



### 1.5 Polynomial Regression

#### Overview
Uses polynomial combinations of existing features to allow a linear model to capture nonlinear relationships.

Polynomial regression is **still linear regression in its coefficients**.

**Scaling is usually recommended after polynomial feature generation.**

#### Syntax

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

poly_reg = Pipeline([
    ("poly", PolynomialFeatures(degree=2, include_bias=False)),
    ("model", LinearRegression())
])
```

#### Parameters

| Parameter | Effect |
|---|---|
| `degree` | Maximum polynomial degree |
| `include_bias` | Adds constant feature |
| `interaction_only` | Creates only interaction terms |

#### Tuning strategy

Usually test small degrees:

```python
param_grid = {
    "poly__degree": [2, 3, 4]
}
```

Avoid very large degrees.

#### Advantages
- Simple way to model nonlinear relationships
- Retains interpretability for small models
- Works with standard linear-model tools

#### Disadvantages
- Number of features grows rapidly
- High-degree models easily overfit
- Poor extrapolation
- Can become computationally expensive

### 1.8 Gradient Boosting Regressor


#### Overview

#### Additional Notes
##### Classification Criteria
- $C$: number of classes
- $p_i$ probability of element belonging to class $i$

*Gini Impurity*: favours splits that make the child nodes purer (simpler)
$$
\mathrm{Gini} = 1 - \sum_{i=1}^{C}(p_i)^2
$$

*Entropy*: favours splits that gain information
$$
\mathrm{Entropy} = -\sum_{i=1}^C p_i \log_2 p_i
$$

| `C` | Strength of penalty for classification errors |
| `class_weight` | Weight assigned to classes |
| `loss` | Loss function |

#### Tuning strategy

```python
param_grid = {
    "C": [0.001, 0.01, 0.1, 1, 10, 100]
}
```

Large `C`:
- fits training data more strongly
- weaker regularization

Small `C`:
- wider margin
- stronger regularization

#### Advantages
- Effective in high dimensions
- Fast compared with kernel SVM
- Strong classifier for sparse data

#### Disadvantages
- Linear decision boundary
- Requires scaling
- `LinearSVC` does not directly provide probabilities

### 2.8 Kernel SVM

#### Overview
Uses the kernel trick to construct nonlinear decision boundaries.

Particularly powerful for small-to-medium datasets with complicated class boundaries.

**Scaling essential.**

#### Syntax

```python
from sklearn.svm import SVC

svc = SVC(
    C=1.0,
    kernel="rbf",
    gamma="scale",
    probability=False
)
```

#### Parameters

| Parameter | Effect |
|---|---|
| `C` | Penalty for classification errors |
| `kernel` | Kernel function |
| `gamma` | Influence radius of individual samples |
| `degree` | Polynomial kernel degree |

#### Tuning strategy

```python
param_grid = {
    "C": [0.1, 1, 10, 100],
    "gamma": ["scale", 0.001, 0.01, 0.1, 1],
    "kernel": ["rbf", "linear"]
}
```

#### Advantages
- Powerful nonlinear classifier
- Flexible
- Effective in high-dimensional spaces

#### Disadvantages
- Slow on large datasets
- Requires scaling
- Sensitive to `C` and `gamma`
- Less interpretable

### 2.9 Gaussian Naive Bayes

#### Overview
Probabilistic classifier based on Bayes' theorem, assuming features are conditionally independent and Gaussian within each class.

Useful as a very fast baseline for continuous features.

**Scaling generally not required.**

#### Syntax

```python
from sklearn.naive_bayes import GaussianNB

gnb = GaussianNB(
    var_smoothing=1e-9
)
```

#### Parameters

| Parameter | Effect |
|---|---|
| `var_smoothing` | Numerical stability / variance smoothing |

#### Advantages
- Extremely fast
- Works with small datasets
- Produces probabilities
- Very few parameters

#### Disadvantages
- Strong independence assumption
- Gaussian assumption may be unrealistic
- Often inferior when features are strongly dependent

#### Additional Notes

Bayes' theorem:

$$
P(C|X)=\frac{P(X|C)P(C)}{P(X)}
$$


### 2.10 Multinomial Naive Bayes

#### Overview
Naive Bayes model designed for discrete count-like features.

Commonly used for **text classification**, particularly word-count or TF-IDF representations.

**Do not standardize into negative values.**

#### Syntax

```python
from sklearn.naive_bayes import MultinomialNB

mnb = MultinomialNB(
    alpha=1.0
)
```

#### Parameters

| Parameter | Effect |
|---|---|
| `alpha` | Additive smoothing strength |

#### Advantages
- Extremely fast
- Excellent simple text baseline
- Effective for sparse matrices

#### Disadvantages
- Designed for non-negative features
- Independence assumption
- Not appropriate for general continuous features
